```python
def calculate_balance(shifts, expenses, target_month):
    balance = calculate_monthly_income(shifts,target_month) - calculate_monthly_expense(expenses,target_month)
    print(f"{target_month}月月の収支: {balance}円")

    if balance<0:
        hourly_wage = 1200
        required_hours = abs(balance)/hourly_wage
        print(f"必要なシフト: {required_hours:.1f}時間")
    else:
        print(f"必要なシフト: ないです")

    return 
    
```
